## Directed Graphs

```python
DG = nx.DiGraph()
DG.add_weighted_edges_from([(1, "meep", 100), ("meep", 3, 1), (3, 1, 0.5)])
print(DG.out_degree(1, weight="weight"))
print(DG.in_degree(1, weight="weight"))
```

```python
list(DG.successors(1))
```

```python
list(DG.predecessors(1))
```

```python
list(DG.neighbors(1))
```
