```python
from collections import defaultdict
counts = defaultdict(int)
for sport in ["Baseball", "Football", "Soccer", "Lacrosse", "Volleyball"]:
    counts[sport] += 1
```

```python
from collections import Counter
sports = ["Lacrosse", "Volleyball", "Baseball", "Football", "Soccer", "Lacrosse", "Volleyball"]
counts = Counter(sports)
print(counts.most_common(3))
```

```python
from collections import deque
q = deque([1, 1, 2, 3, 5, 8, 13, 21])
q.appendleft(0)
print(q)
q.pop()
q.popleft()
print(q)
```

```python
import heapq
bids = [29.2, 29.4, 27.0, 25.3, 32]
heapq.heapify(bids)
print(bids)
heapq.heappush(bids, 24.7)
print(bids)
heapq.heappop(bids)
print(bids)
```
